```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train = pd.read_csv("/kaggle/input/home-data-for-ml-course/train.csv")
test = pd.read_csv("/kaggle/input/home-data-for-ml-course/test.csv")

# print("Train data preview:")
# print(train.head())
# print("\nTrain data info:")
# print(train.info())
```

```python
def fill_missing_values(df):

    df = df.copy()
    
    # Fill numeric columns with median
    numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
    for col in numeric_cols:
        df[col] = df[col].fillna(df[col].median())
    
    # Fill categorical columns with mode
    categorical_cols = df.select_dtypes(include=['object']).columns
    for col in categorical_cols:
        df[col] = df[col].fillna(df[col].mode()[0])
    
    return df


train = fill_missing_values(train)
test = fill_missing_values(test)
print("Train columns:", train.columns)

train.drop(['Id'], axis=1, inplace=True, errors='ignore')

if 'Id' in test.columns:
    test_ids = test['Id']
    test.drop(['Id'], axis=1, inplace=True)
# else:
#     test_ids = pd.Series(range(len(test)), name='Id')

#print("Columns in train data:", train.columns.tolist())

if 'SalePrice' in train.columns:
    y = train['SalePrice']
    train.drop(['SalePrice'], axis=1, inplace=True)
# else:
#     raise KeyError("The 'SalePrice' column is missing from the training data.")
```

```python
all_data = pd.concat([train, test], axis=0)
all_data = pd.get_dummies(all_data)

X_train = all_data.iloc[:train.shape[0], :]
X_test = all_data.iloc[train.shape[0]:, :]

model = RandomForestRegressor(n_estimators=100, random_state=42)

cv_scores = cross_val_score(model, X_train, y, cv=5, scoring='neg_mean_squared_error')
rmse_scores = np.sqrt(-cv_scores)
print("Cross-validated RMSE: {:.2f}".format(rmse_scores.mean()))

model.fit(X_train, y)

predictions = model.predict(X_test)

submission = pd.DataFrame({
    "Id": test_ids,
    "SalePrice": predictions
})

submission.to_csv("submission.csv", index=False)
print("Submission file 'submission.csv' created successfully!")
```
